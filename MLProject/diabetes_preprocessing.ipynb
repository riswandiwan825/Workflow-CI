Notebook ini melakukan **muat data, analisis data eksploratori (EDA), analisis korelasi fitur, pembagian dataset, dan skaling fitur** pada dataset **Diabetes Health Classification** sebelum dimasukkan ke dalam pipeline pelatihan MLflow.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Unduh & Muat dataset Diabetes
URL_DATASET = 'https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv'
df_diabetes = pd.read_csv(URL_DATASET)
print(f'Data Diabetes Loading Selesai! Dimensi: {df_diabetes.shape}')
df_diabetes.head()
```

```python
print('Informasi Kolom & Tipe Data:')
df_diabetes.info()
```

```python
# Visualisasi Distribusi Target Kelas Outcome (0: Non-Diabetic, 1: Diabetic)
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.countplot(x='Outcome', data=df_diabetes, palette='Set2', ax=ax)
ax.set_title('Distribusi Kelas Diabetes (0: Non-Diabetic, 1: Diabetic)', fontsize=12, fontweight='bold')
ax.set_xlabel('Kelas Outcome')
ax.set_ylabel('Jumlah Sampel')
plt.tight_layout()
plt.show()
```

```python
print('Distribusi Target Outcome:')
print(df_diabetes['Outcome'].value_counts())
```

```python
X = df_diabetes.drop(columns=['Outcome'])
y = df_diabetes['Outcome']

# Split 80% Latih / 20% Uji
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Skaling Fitur via StandardScaler
scaler = StandardScaler()
X_tr_sc = scaler.fit_transform(X_train)
X_te_sc = scaler.transform(X_test)

# Simpan berkas CSV untuk tahap pelatihan MLflow
pd.DataFrame(X_tr_sc, columns=X.columns).to_csv('X_train.csv', index=False)
pd.DataFrame(X_te_sc, columns=X.columns).to_csv('X_test.csv', index=False)
pd.DataFrame(y_train).to_csv('y_train.csv', index=False)
pd.DataFrame(y_test).to_csv('y_test.csv', index=False)

print('Proses Preprocessing & Pembagian Dataset Selesai!')
print(f'  Data Latih : {len(X_train):,} sampel')
print(f'  Data Uji   : {len(X_test):,} sampel')
print('  File tersimpan: X_train.csv, X_test.csv, y_train.csv, y_test.csv')
```
