## 📊 Perkenalan Dataset

- **Nama Dataset** : Red Wine Quality Dataset

- **URL Berkas**   : `https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv`  

### 📝 Gambaran Singkat Isi Data:
Dataset ini berisi data hasil pengujian fisikokimia dari varian anggur merah (*Vinho Verde*) Portugis.
Dataset memiliki **1.599 sampel** dengan **11 fitur masukan (fisikokimia)** dan **1 kolom target (`quality`)**:

| No | Fitur | Deskripsi |
|----|-------|-----------|
| 1  | `fixed acidity` | Kadar asam fixed (asam tartarat) |
| 2  | `volatile acidity` | Kadar asam volatil (asam asetat, berkaitan dengan rasa cuka) |
| 3  | `citric acid` | Kadar asam sitrat (memberikan kesegaran) |
| 4  | `residual sugar` | Sisa gula setelah proses fermentasi selesai |
| 5  | `chlorides` | Kadar garam dalam wine |
| 6  | `free sulfur dioxide` | Bentuk SO2 bebas yang mencegah pertumbuhan mikroba |
| 7  | `total sulfur dioxide` | Total kadar SO2 bebas dan terikat |
| 8  | `density` | Kepadatan/massa jenis anggur terlarut |
| 9  | `pH` | Tingkat keasaman skala 0 (sangat asam) s.d. 14 (basa) |
| 10 | `sulphates` | Aditif garam mineral yang berkontribusi pada kadar SO2 |
| 11 | `alcohol` | Persentase kadar alkohol (%) |
| **Target** | `quality` | Skor kualitas dari penguji ahli (skala 0 s.d. 10) |

Pada tahap pra-pemrosesan ini, kolom target `quality` ditransformasikan menjadi **klasifikasi biner**:
- `1` (**Kualitas Tinggi / Good**) : jika nilai `quality > 6`
- `0` (**Kualitas Standar / Bad**) : jika nilai `quality <= 6`

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Gunakan dataset mentah lokal jika ada, atau unduh langsung dari UCI ML Repository
FILE_RAW = 'winequality-red_raw.csv'
URL_DATASET = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv'

if os.path.exists(FILE_RAW):
    df_wine = pd.read_csv(FILE_RAW, sep=';')
else:
    df_wine = pd.read_csv(URL_DATASET, sep=';')

print(f'Data Loading Selesai! Dimensi mentah: {df_wine.shape}')
df_wine.head()
```

```python
print('Informasi Kolom & Tipe Data:')
df_wine.info()
```

```python
# Visualisasi Distribusi Target Kualitas Asli
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.countplot(x='quality', data=df_wine, palette='viridis', ax=ax)
ax.set_title('Distribusi Nilai Kualitas Wine (Data Asli)', fontsize=12, fontweight='bold')
ax.set_xlabel('Nilai Kualitas (3-8)')
ax.set_ylabel('Jumlah Sampel')
plt.tight_layout()
plt.show()
```

```python
# Konversi target menjadi biner (1: quality > 6, 0: quality <= 6)
df_wine['quality'] = (df_wine['quality'] > 6).astype(int)
print('Distribusi Target Biner (1 = Kualitas Tinggi, 0 = Kualitas Rendah):')
print(df_wine['quality'].value_counts())
```

```python
X = df_wine.drop(columns=['quality'])
y = df_wine['quality']

# Split 80% Latih / 20% Uji
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Skaling Fitur via StandardScaler
scaler = StandardScaler()
X_tr_sc = scaler.fit_transform(X_train)
X_te_sc = scaler.transform(X_test)

# Simpan berkas CSV untuk tahap pelatihan MLflow
pd.DataFrame(X_tr_sc, columns=X.columns).to_csv('X_train.csv', index=False)
pd.DataFrame(X_te_sc, columns=X.columns).to_csv('X_test.csv', index=False)
pd.DataFrame(y_train).to_csv('y_train.csv', index=False)
pd.DataFrame(y_test).to_csv('y_test.csv', index=False)

print('Proses Preprocessing & Pembagian Dataset Selesai!')
print(f'  Data Latih : {len(X_train):,} sampel')
print(f'  Data Uji   : {len(X_test):,} sampel')
print('  File tersimpan: X_train.csv, X_test.csv, y_train.csv, y_test.csv')
```
